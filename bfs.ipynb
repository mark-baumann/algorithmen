# 🌊 BFS — Breitensuche

**O(V+E)** — kürzester Pfad in ungewichteten Graphen.

---
## 📚 Idee
Erkunde Ebene für Ebene. Perfekt für kürzeste Pfade in ungewichteten Graphen.

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
from collections import deque

def bfs(graph: dict, start) -> list:
    """Breitensuche — kürzester Pfad in ungewichtetem Graphen."""
    visited = {start}
    queue = deque([start])
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return order

# Beispiel-Graph
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "F"],
    "F": ["C", "E"],
}

print(f"BFS ab A: {bfs(graph, 'A')}")
print(f"→ Ebene 0: A, Ebene 1: B,C, Ebene 2: D,E,F")
```

## 🎯 Zusammenfassung

| Eigenschaft | Wert |
|------------|------|
| Laufzeit | O(V+E) |
| Anwendung | Kürzester Pfad (ungewichtet), Level-Order |
