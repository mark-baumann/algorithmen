# 🔎 Lineare Suche

**O(n)** — funktioniert auf unsortierten Daten.

---
## 📚 Idee
Gehe jedes Element der Reihe nach durch, bis du das Ziel findest.

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
def linear_search(arr: list, target) -> int:
    """O(n) — funktioniert auf unsortierten Daten."""
    for i, x in enumerate(arr):
        if x == target:
            return i
    return -1

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Linear Search 22: Index {linear_search(arr, 22)}")
print(f"Linear Search 99: Index {linear_search(arr, 99)} (nicht gefunden)")
```

## 🎯 Zusammenfassung

| Eigenschaft | Wert |
|------------|------|
| Laufzeit | O(n) |
| Voraussetzung | Keine |
