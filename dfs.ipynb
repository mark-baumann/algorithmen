# 🔍 DFS — Tiefensuche

**O(V+E)** — Zyklen-Erkennung, Topologische Sortierung.

---
## 📚 Idee
Gehe so tief wie möglich, dann zurück (Backtracking).

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
def dfs(graph: dict, start, visited: set | None = None) -> list:
    """Tiefensuche — rekursiv."""
    if visited is None:
        visited = set()
    visited.add(start)
    order = [start]
    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            order.extend(dfs(graph, neighbor, visited))
    return order

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "F"],
    "F": ["C", "E"],
}

print(f"DFS ab A: {dfs(graph, 'A')}")
print(f"→ Geht erst tief (A→B→D), dann zurück")
```

## 🎯 Zusammenfassung

| Eigenschaft | Wert |
|------------|------|
| Laufzeit | O(V+E) |
| Anwendung | Zyklen finden, Topologische Sortierung |
