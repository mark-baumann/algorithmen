# ⚡ Quick Sort

**O(n log n)** — Divide & Conquer.

---
## 📚 Idee
Wähle ein Pivot-Element, teile in "kleiner" und "größer", sortiere rekursiv.

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
def quick_sort(arr: list) -> list:
    """O(n log n) average — Divide & Conquer."""
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original: {arr}")
print(f"Quick:    {quick_sort(arr)}")
```
