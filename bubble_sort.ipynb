# 🫧 Bubble Sort

**O(n²)** — der einfachste Sortieralgorithmus.

---
## 📚 Idee
Vergleiche benachbarte Elemente und vertausche sie, wenn sie in der falschen Reihenfolge sind. Wiederhole, bis nichts mehr getauscht wird.

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
def bubble_sort(arr: list) -> list:
    """O(n²) — einfach, aber langsam. Gut zum Lernen."""
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# Test
arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original: {arr}")
print(f"Bubble:   {bubble_sort(arr)}")
```
