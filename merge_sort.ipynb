# 🔀 Merge Sort

**O(n log n)** — stabil, Divide & Conquer.

---
## 📚 Idee
Teile das Array in der Mitte, sortiere beide Hälften rekursiv, führe sie zusammen.

> 💡 **Google Colab:** Datei → In Drive speichern → mit GPU/TPU ausführen.


```python
def _merge(left: list, right: list) -> list:
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(arr: list) -> list:
    """O(n log n) — stabil, gut für Linked Lists."""
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return _merge(left, right)

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original: {arr}")
print(f"Merge:    {merge_sort(arr)}")
```

## 🎯 Zusammenfassung

| Eigenschaft | Wert |
|------------|------|

| Laufzeit (worst) | O(n log n) |
| Stabil | ✅ |
| In-Place | ❌ |
